## Inheritance
* A class can inherit from another class.
* Child class inherits methods and attributes of Parent class.
* Object of Child class can access non private attributes and non private methods of Parent Class.
* Child class can override methods or attributes
* Parent has no access to Child Class.

### Why/Benefit of Inheritance?
* To prevent code repeatation
* Code reusability


```python
class User:
  def __init__(self):
    self.name = 'shreyas'

  def login(self):
    print('login')

class Student(User):            # syntax of inheritance
  def register(self):
    print('successful')
```

* Always child constructor is executed first.

```python
u1 = User()
s1 = Student()
```

```python
s1.name # object of child class has access to attribute of parent class
```

```python
s1.login()  # object of child class has access to method of parent class
```

```python
print(s1.name)
```

### What if Constructor is not present in Child class?

* When we create object of child class,
* It checks for the **constructor** in the **Child** class.
* If not present it will execute the **constructor** of **Parent** class.
* Inherit the attributes/ code in the parent constructor.

```python
# Child Class not having Constructor

class Phone:
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Bought a phone")

class SmartPhone(Phone):
  pass
```

```python
s1 = SmartPhone(20000, 'RedMi', 13)

# Child constructor not present
# So parent constructor is being called/executed.
```

```python
print(s1.brand)
print(s1.price)
print(s1.camera)

# As Parent constructor is being called/ executed
# So the attributes in the Parent constructor were inherited by Child Class
```

```python
s1.buy()

# Non-Private Method of parent class is inherited by child and executed upon calling
```

### What if Constructor is present in Child class?

* If constructor is present in Child class,
* It will not execute the constructor in Parent class.
* It will never call constructor of Parent Class.
* So the attributes?code defined in the constructor of Parent class will never be inherited by Child class.
* It happens due to **Method Overriding**.




```python
# Constructor present in Child Class.

class Phone:
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.price = price
    self.brand = brand
    self.camera = camera

  # def buy(self):
  #   print("Bought a phone")

class SmartPhone(Phone):
  def __init__(self, os, ram):
    self.os = os
    self.ram = ram
    print("Inside SmartPhone Constructor")
```

```python
s2 = SmartPhone("Android", 128)

# Constructor is Present in both Child as well as Parent
# So Parent constructor is never called or executed ---- METHOD OVERRIDING
# So the attributes present in Parent Constructor were never inherited by Child Class.
```

```python
print(s2.brand)
print(s2.price)
print(s2.camera)

# As Parent constructor is never being called/ executed
# So the attributes in the Parent constructor were never inherited by Child Class
# So Program will crash -- Throws error
```

### Method Overriding
* When methods of same name is present in both Parent as well as Child class.
* Parent method will never be called or executed
* Hence the child method will be executed.

```python
# Method Overriding.

class Phone:
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Im in Parent Method")

class SmartPhone(Phone):
  def buy(self):
    print("Im in child method")
```

```python
s5 = SmartPhone(10000, 'Android', 56)
```

```python
s5.buy()

# As the Parent class also contains method with same name
# It got overrided by the method in Child class
# Hence the 'buy' method in Parent class was never executed or called
# And so the 'buy' method in Child class was executed
```

### How to Prevent **method overriding** of child constructor?

* To prevent this we use **super** key word.

### Super Key Word
* Prevents Method Overriding.
* Way to access parent methods with same name in case of method overriding.
* It cannot access attributes, only methods.
* Syntax = super().method_name()
* Used inside Child Class always.

```python
# Super Key Word.

class Phone:
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Im in Parent Method")

class SmartPhone(Phone):
  def buy(self):
    super().buy()
    print("Im in child method")
    # super().buy()   # Syntax to call parent method with same name
```

```python
s6 = SmartPhone(10000, 'Android', 56)
```

```python
s6.buy()
```

```python
# Example on super() -- constructor

class Phone:
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Im in Parent Method")

class SmartPhone(Phone):
  def __init__(self, price, brand, camera, os, ram):
    print("Im inside SmartPhone Constructor")
    super().__init__(price, brand, camera)
    self.os = os
    self.ram = ram
    print("Im inside smart Phone constructor")

  def buy(self):
    super().buy()
    print("Im in child method")
    # super().buy()   # Syntax to call parent method with same name
```

```python
s7 = SmartPhone(10000, 'Apple', 56, 'IOS', 64)
```

### How to Access Private attributes and methods of Parent class?
#### By using non private Getter method in Parent Class
* As we know private attributes and methods of Parent Class are not accessible or inherited by Child Class.
* But by using non private getter method in Parent class, It can print or return private attribute
* As getter method is non private, child class has access to all the code inside it.
* But if that getter method method private and we are calling it, it wont execute.
* Because it wont be inherited by child class.

### Conclusion: Child Cass cant inherit Private attributes and Methods of Parent Class.

```python
# Child class can't access private attributes and methods of the Parent Class.
# Private attributes

class Phone:
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def display(self):
    return self.__price

class SmartPhone(Phone):
  def check(self):
    print(self.__price)
```

```python
s3.check()

# Check method of Child class is trying to print private attribute of parent class
# So it will throw error
```

```python
# Child class can't access private attributes and methods of the Parent Class.
# Private methods

class Phone:
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def __display(self):
    return self.__price

class SmartPhone(Phone):
  def check(self):
    print(self.__price)
```

### Accessing Private Attributes through Getter methods.

```python
# Accessig Private attributes through getter method

class Phone:
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def display(self):          # getter method
    return self.__price

class SmartPhone(Phone):
  def check(self):
    print("Inside Child Class")
```

## Types of Inheritance
* 5 Types of Inheritance

### 1. Single Inheritance
* One Parent and One Child.
* Child inherits from the Parent

```python
# Single Inheritance

class Phone:
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Bought a phone")

class SmartPhone(Phone):
  pass

s1 = SmartPhone(20000, 'RedMi', 13)
s1.buy()
```

### 2. Multilevel Inheritance
* A chain of classes.
* It can be infinitely long.
* A class can be a Parent and Child at the same time.
* A child can inherit everything from its all above parents.
* A child can have only one parent
* A Parent can have only one child
* Like this there are many child and parents

```python
# Multilevel Inheritance

class Product:                                           # grandparent class
  def review(self):
    print("Product Customer review")

class Phone(Product):                                   # parent class
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Bought a phone")

class SmartPhone(Phone):                                # child class
  pass

s1 = SmartPhone(20000, 'RedMi', 13)
s1.buy()       # method of parent -- accessible by child
s1.review()    # method of grandfathaer -- accessible by child
```

### 3. Hierarchial Inheritance
* One Parent has multiple children.
* Every child has access to parent's non private methods and non private attributes.

```python
# Hierarchial Inheritance

class Phone:                                          # Parent class
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Bought a phone")

class SmartPhone(Phone):                               # Child 1 class
  pass

class IPhone(Phone):                                   # Child 2 class
  pass

s1 = SmartPhone(20000, 'RedMi', 13)
i1 = IPhone(30000, 'oppo', 26)
s1.buy()
i1.buy()

# both child has access to parent methods
```

### 4. Multiple Inheritance
* A child can have multiple Parents
* In case both parents have same method name, it will execute the method of that class which we have passed as first parent.
* It is called Method Resolution Order.

```python
# Multiple Inheritance

class Phone:                                   # 1st Parent class
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Bought a phone")

class Product:                                 # 2nd Parent class
  def review(self):
    print("Product Customer review")

class SmartPhone(Phone, Product):              # Child of both Parent
  pass

s1 = SmartPhone(20000, 'RedMi', 13)
s1.buy()
s1.review()

# child has access to methods(non-private) and attributes(non-private) of both parents
```

```python
# Diamond Problem

class Phone:                                   # 1st Parent class
  def __init__(self, price, brand, camera):
    print("Inside Phone Constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print("Bought a phone")

class Product:                                 # 2nd Parent class
  def buy(self):
    print("Bought a product")

class SmartPhone(Phone, Product):              # Child of both Parent
  pass

s1 = SmartPhone(20000, 'RedMi', 13)
s1.buy()
```

## Polymorphism
* Means Multiple Forms
* Method Overriding
* Method Overloading

### Method Overloading
* When there are methods of same name in a single class.
* But behaviour is different depending on the input.
* Output is different.

#### Why use Method Overloading/Benefit?
* Code is more clean while reading
* To avoid unnecessary long names.
* Based on input passed it will decide which method to execute upon calling.

#### Conclusion: But in Python it wont work upon calling if name are same.

```python
class Shape:
  def area(self, radius):
    return 3.14 * radius * radius

  def area(self, l, b):
    return l*b

# name same but behaviour different --- based on input
```

```python
sh = Shape()
sh.area(4)
sh.area(2,3)

# will throw error
```

### How to Implement method overloading in python then?

```python
class Shape:
  def area(self, a, b = 0):
    if b == 0:
      return 3.14 * a* a
    else:
      return a * b

sh = Shape()
print(sh.area(3))
print(sh.area(2,3))
```

## Abstraction
* Means hidden.
* We can apply constraints on the lower classes.
* Abstract Class - Has atleast one abstract method.
* Abstract Method - Has no code written in it but we can use them and has decorator on it @abstractmethod.
* Concrete methods - Till now we have made concrete functions.

### We can't make Object of Abstract Class.

```python
from abc import ABC, abstractmethod
class BankApp(ABC):        # This is an abstract class -- as it inherits from ABC Class.
  def database(self):
    print("Connected to Database")

  @abstractmethod          # this is decorator -- by applying this, the method becomes abstract
  def security(self):
    pass

# It inherits from ABC Class
# Has atleast one abstract method
# So this class BankApp is an Abstract Class
```

```python
class MobileApp(BankApp):
  def mobile_login(self):
    print("Login into mobile")

  # constraint to use parent method(database) -- there should be security method in Child Class
```

```python
mob = MobileApp()

# Will throw error -- Can't instantiate abstract class MobileApp with abstract method security
```
